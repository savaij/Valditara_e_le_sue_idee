# Casi anomali da controllare nei dataset originali

Ogni sezione mostra le righe **grezze** (`data_raw/`) di uno dei fenomeni emersi dai controlli qualità.

```python
import pandas as pd
from pathlib import Path

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 250)

RAW = Path('../data_raw')
SUFFISSI = {'statale': ('STA', 'STAT'), 'paritaria': ('PAR', 'PAR')}


def leggi(nome):
    return pd.read_csv(RAW / nome, dtype=str, keep_default_na=False, encoding='utf-8-sig')


def carica(prefisso, gestione, anagrafe=False):
    suffisso = SUFFISSI[gestione][1 if anagrafe else 0]
    df = leggi(f'{prefisso}{suffisso}20242520250831.csv')
    df.insert(0, 'gestione', gestione)
    return df


# Flusso cittadinanza, flusso classi/studenti, anagrafe standard e autonome (statali + paritarie)
cit = pd.concat([carica('ALUITASTRACIT', g) for g in SUFFISSI], ignore_index=True)
cla = pd.concat([carica('ALUCORSOINDCLA', g) for g in SUFFISSI], ignore_index=True)
ana_std = pd.concat([carica('SCUANAGRAFE', g, anagrafe=True) for g in SUFFISSI], ignore_index=True)
ana_aut = pd.concat([carica('SCUANAAUT', g, anagrafe=True) for g in SUFFISSI], ignore_index=True)
ana_std['fonte'] = 'standard'
ana_aut['fonte'] = 'autonome'
ana = pd.concat([ana_std, ana_aut], ignore_index=True)

for col in ['ALUNNI', 'ALUNNICITTADINANZAITALIANA', 'ALUNNICITTADINANZANONITALIANA',
            'ALUNNICITTADINANZANONITALIANAPAESIUE', 'ALUNNICITTADINANZANONITALIANAPAESINONUE']:
    cit[col] = pd.to_numeric(cit[col].replace('', '0'))
for col in ['CLASSI', 'ALUNNIMASCHI', 'ALUNNIFEMMINE']:
    cla[col] = pd.to_numeric(cla[col].replace('', '0'))
cla['ALUNNI_CLASSI'] = cla['ALUNNIMASCHI'] + cla['ALUNNIFEMMINE']
cit['QUOTA'] = cit['ALUNNICITTADINANZANONITALIANA'] / cit['ALUNNI']
cit['SOPRA_30'] = 10 * cit['ALUNNICITTADINANZANONITALIANA'] > 3 * cit['ALUNNI']

# Anagrafe ridotta (primo codice, come in analyze.py) da unire alle righe grezze
ANA_COLS = ['CODICESCUOLA', 'DENOMINAZIONESCUOLA', 'CODICEISTITUTORIFERIMENTO', 'DESCRIZIONECOMUNE',
            'PROVINCIA', 'DESCRIZIONECARATTERISTICASCUOLA', 'DESCRIZIONETIPOLOGIAGRADOISTRUZIONESCUOLA', 'fonte']
ana_1 = ana.drop_duplicates('CODICESCUOLA', keep='first')[ANA_COLS]
citx = cit.merge(ana_1, on='CODICESCUOLA', how='left')

len(cit), len(cla), len(ana)
```

```python
def mostra(codice):
    """Tutte le righe grezze di un plesso: anagrafe, cittadinanza, classi."""
    print('--- ANAGRAFE')
    display(ana[ana['CODICESCUOLA'].eq(codice)].T)
    print('--- CITTADINANZA')
    display(cit[cit['CODICESCUOLA'].eq(codice)].sort_values(['ORDINESCUOLA', 'ANNOCORSO']))
    print('--- CLASSI/STUDENTI')
    display(cla[cla['CODICESCUOLA'].eq(codice)].sort_values(['ORDINESCUOLA', 'ANNOCORSOCLASSE']))
```

## 1. Unità = anno di corso, non classe

Il flusso classi dà solo il **numero** di classi per anno di corso, non la composizione di ciascuna. Qui si vede quante sezioni vengono sommate nelle unità sopra soglia.

```python
chiave_cit = ['gestione', 'CODICESCUOLA', 'ORDINESCUOLA', 'ANNOCORSO']
uc = cit.merge(
    cla.rename(columns={'ANNOCORSOCLASSE': 'ANNOCORSO'})[chiave_cit + ['CLASSI', 'ALUNNI_CLASSI']],
    on=chiave_cit, how='left',
)
```

```python
chiave_cit = ['gestione', 'CODICESCUOLA', 'ORDINESCUOLA', 'ANNOCORSO']
uc = cit.merge(
    cla.rename(columns={'ANNOCORSOCLASSE': 'ANNOCORSO'})[chiave_cit + ['CLASSI', 'ALUNNI_CLASSI']],
    on=chiave_cit, how='left',
)
uc['ALUNNI_PER_CLASSE'] = uc['ALUNNI'] / uc['CLASSI']

# Distribuzione del numero di classi per unità, separando unità sopra/sotto soglia
display(pd.crosstab(pd.cut(uc['CLASSI'], [0, 1, 2, 4, 8, 12, 100]), uc['SOPRA_30'], margins=True))

# Unità con molte classi appena sotto soglia (25-30%): qui è probabile che qualche sezione sia sopra il 30%
uc[(uc['CLASSI'] >= 4) & uc['QUOTA'].between(0.25, 0.30)].sort_values('CLASSI', ascending=False).head(20)
```

## 6. Pluriclassi (anno di corso 7) e differenze tra i due flussi

Nel flusso classi, gli alunni delle pluriclassi stanno sotto `ANNOCORSOCLASSE = 7`; nel flusso cittadinanza sono distribuiti sui loro anni di corso effettivi.

```python
cla[cla['ANNOCORSOCLASSE'].eq('7')].groupby(['gestione', 'ORDINESCUOLA']).agg(
    righe=('CLASSI', 'size'), classi=('CLASSI', 'sum'), alunni=('ALUNNI_CLASSI', 'sum'))
```

```python
# Confronto tra i due flussi per plesso: il totale deve tornare includendo l'anno 7
tot_cit = cit.groupby(['gestione', 'CODICESCUOLA', 'ORDINESCUOLA'])['ALUNNI'].sum()
tot_cla = cla.groupby(['gestione', 'CODICESCUOLA', 'ORDINESCUOLA'])['ALUNNI_CLASSI'].sum()
confronto = pd.concat([tot_cit, tot_cla], axis=1).fillna(0)
confronto['delta'] = confronto['ALUNNI'] - confronto['ALUNNI_CLASSI']
print('plessi con totale diverso:', (confronto['delta'] != 0).sum())
confronto[confronto['delta'] != 0]
```
